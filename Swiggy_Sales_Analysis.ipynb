```python
import pandas as pd
import matplotlib.pyplot as plt

# Assuming '/content/swiggy_data.xlsx' is the correct path for your data
df = pd.read_excel("/content/swiggy_data.xlsx")

#Daily Trend
df["DayName"] = pd.to_datetime(df["Order Date"]).dt.day_name()

daily_revenue = (
    df.groupby("DayName")["Price (INR)"]
      .sum()
      .reindex(["Monday", "Tuesday", "Wednesday", "Thursday",
                "Friday", "Saturday", "Sunday"])
)

plt.figure(figsize=(10, 5))
plt.bar(daily_revenue.index, daily_revenue.values)

plt.title("Daily Revenue Trend (Mon-Sun)")
plt.xlabel("Day")
plt.ylabel("Revenue (INR)")
plt.xticks(rotation=30)

plt.show()
```

```python
#Total Sales by Food Type(Veg vs Non Veg)

non_veg_keywords = [
    "chicken" , "eggs" , "fish" , "mutton" ,
    "prawn" , "biryani" , "kabab" , "kebab" ,
    "non-veg" , "non veg"
]

df["Food Category"] = np.where(
    df["Dish Name"].str.lower().str.contains('|'.join(non_veg_keywords)),
    "Non-Veg" , "Veg"
)
```

```python
food_revenue = (
    df.groupby("Food Category")["Price (INR)"]
      .sum()
      .reset_index()
)
```

```python
fig = px.pie(
    food_revenue,
    names="Food Category",
    values="Price (INR)",
    hole = 0.5,
    title="Total Sales by Food Type"
)
fig.update_traces(
    textinfo="percent+label",
    pull=[0.05,0]
)
fig.update_layout(
    height =500,
    margin=dict(t=40,b=40,l=40,r=40)
)
fig.show()
```

```python
#Total sales by State
fig = px.bar(
    df.groupby("State" , as_index=False)["Price (INR)"].sum().sort_values("Price (INR)", ascending=False),
    y="State",
    x="Price (INR)",
    orientation="h",
    title="Total Sales by State"
)

fig.update_layout(height=600 , yaxis=dict(autorange = "reversed"))
fig.show()
```

```python
#Quarterly Performance Summary
df["Order Date"] = pd.to_datetime(df["Order Date"])
df["Quarter"] = df["Order Date"].dt.to_period("Q").astype(str)

quarterly_summary = (
    df.groupby("Quarter" , as_index=False)
      .agg(
          Total_Sales=("Price (INR)" , "sum"),
          Avg_rating=("Rating" , "mean"),
          Total_orders=("Price (INR)" , "count")
          ).sort_values("Quarter")
)
quarterly_summary["Total_Sales"] = quarterly_summary["Total_Sales"].apply(lambda x : round(x,2))
quarterly_summary["Avg_rating"] = quarterly_summary["Avg_rating"].apply(lambda x : round(x,2))
quarterly_summary
```

```python
#Top 5 Cities by Sales
top_5_cities = (
    df.groupby("City")["Price (INR)"]
    .sum()
    .nlargest(5)
    .sort_values()
    .reset_index()
)

fig = px.bar(
    top_5_cities,
    y="City",
    x="Price (INR)",
    orientation="h",
    title="Top 5 Cities by Sales",
    color_discrete_sequence=["red"]
)
fig.update_layout(height=600)
fig.show()
```

```python
df["DayOfWeek"] = df["Order Date"].dt.day_name()

weekly_revenue = (
    df.groupby("DayOfWeek")["Price (INR)"]
      .sum()
      .reindex(["Monday", "Tuesday", "Wednesday", "Thursday",
                "Friday", "Saturday", "Sunday"])
)

plt.figure(figsize=(10, 5))
plt.bar(weekly_revenue.index, weekly_revenue.values)

plt.title("Weekly Revenue Trend")
plt.xlabel("Day of Week")
plt.ylabel("Revenue (INR)")
plt.xticks(rotation=45)

plt.show()
```

### Top 5 Restaurants by Sales

```python
top_5_restaurants = (
    df.groupby("Restaurant Name")["Price (INR)"]
    .sum()
    .nlargest(5)
    .sort_values(ascending=False)
    .reset_index()
)

fig = px.bar(
    top_5_restaurants,
    x="Restaurant Name",
    y="Price (INR)",
    title="Top 5 Restaurants by Sales",
    color_discrete_sequence=px.colors.qualitative.Pastel
)
fig.update_layout(height=500, xaxis={'categoryorder':'total descending'})
fig.show()
```

### Distribution of Ratings

```python
plt.figure(figsize=(10, 6))
sns.histplot(df['Rating'], bins=20, kde=True, color='skyblue')
plt.title('Distribution of Ratings')
plt.xlabel('Rating')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

### Average Rating by Restaurant

```python
avg_rating_by_restaurant = (
    df.groupby("Restaurant Name")["Rating"]
    .mean()
    .nlargest(10)  # Top 10 restaurants by average rating
    .sort_values(ascending=False)
    .reset_index()
)

fig = px.bar(
    avg_rating_by_restaurant,
    x="Restaurant Name",
    y="Rating",
    title="Top 10 Restaurants by Average Rating",
    color_discrete_sequence=px.colors.qualitative.Plotly
)
fig.update_layout(height=500, xaxis={'categoryorder':'total descending'})
fig.show()
```

### Total Sales by Food Category

```python
food_category_sales = (
    df.groupby("Category")["Price (INR)"]
    .sum()
    .nlargest(10) # Display top 10 categories
    .sort_values(ascending=False)
    .reset_index()
)

fig = px.bar(
    food_category_sales,
    x="Category",
    y="Price (INR)",
    title="Top 10 Food Categories by Sales",
    color_discrete_sequence=px.colors.qualitative.D3
)
fig.update_layout(height=500, xaxis={'categoryorder':'total descending'})
fig.show()
```

### Monthly Sales Trend

```python
df['Order Month'] = df['Order Date'].dt.to_period('M').astype(str)
monthly_sales = df.groupby('Order Month')['Price (INR)'].sum().reset_index()

fig = px.line(
    monthly_sales,
    x='Order Month',
    y='Price (INR)',
    title='Monthly Sales Trend',
    markers=True,
    color_discrete_sequence=['purple']
)
fig.update_layout(height=500)
fig.show()
```

### Sales by Location

```python
location_sales = (
    df.groupby("Location")["Price (INR)"]
    .sum()
    .nlargest(10)  # Display top 10 locations by sales
    .sort_values(ascending=False)
    .reset_index()
)

fig = px.bar(
    location_sales,
    x="Location",
    y="Price (INR)",
    title="Top 10 Locations by Sales",
    color_discrete_sequence=px.colors.qualitative.G10
)
fig.update_layout(height=500, xaxis={'categoryorder':'total descending'})
fig.show()
```
